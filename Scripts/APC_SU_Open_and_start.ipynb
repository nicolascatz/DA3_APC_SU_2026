```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat
import os

if os.getcwd().endswith('Scripts'):
    os.chdir('..')
os.getcwd()

listdir=os.listdir('data')
```

```python
for i in range(90,100): #range(len(listdir)):
    data=loadmat('data/'+ listdir[i])
    su=data['single_unit_aligned_sorted']
    list_trial_name=data['list_trial_name']
    N_cond = data['N_cond'].tolist()
    N_cond=N_cond[0]
    
    fig,ax=plt.subplots(figsize=(5,3))
    ax.spy(su[:,:700], markersize=0.5,aspect='auto',color='k',marker='.')
    ax.set_title(f'Unit {listdir[i]}')
    ax.set_xlabel('Time (ms)')
    ax.set_ylabel('Trials')
    ax.legend()
    ax.set_ylim(0, su.shape[0])
    ax.set_aspect('auto')
    ax.grid(True, color='white', linestyle='-', linewidth=0.5)
    ax.set_axisbelow(True)
    ax.axvline(x=200, color='k', linestyle='--', label='Stimulus Onset')
    
    for t in range(1,len(N_cond)):
        ax.axhline(y=N_cond[t], color='k', linestyle='--', label='Condition Boundary')
    
    plt.show()
```
